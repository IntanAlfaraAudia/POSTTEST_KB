### **Dataset yang Digunakan**
Dataset yang digunakan adalah *Palmer Archipelago (Antarctica)* Penguin Data.
Dataset ini berisi data pengamatan penguin yang ditemukan di Kepulauan Palmer,
Antarktika. Data mencatat beberapa informasi mengenai penguin, seperti species,
lokasi pengambilan sampel, identitas individu, tanggal telur, ukuran tubuh,
jenis kelamin, serta hasil pengukuran isotop.

Dataset terdiri dari data numerik dan data kategorikal sehingga dapat digunakan
untuk melakukan eksplorasi data, melihat distribusi, membandingkan kategori,
dan melihat hubungan antar attribute numerik.

### **1. Import Library**

Library yang digunakan adalah Pandas untuk mengolah data, Matplotlib untuk
membuat grafik, dan Seaborn untuk membantu membuat visualisasi statistik.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

### **2. Membaca Dataset**
Dataset dibaca menggunakan Pandas. Setelah data berhasil dibaca,
ditampilkan beberapa data pertama untuk melihat bentuk awal dataset.

```python
df = pd.read_csv("penguins_lter.csv")
df.head()
```

Pada bagian ini dilakukan pengecekan jumlah record, jumlah attribute,
dan nama attribute yang terdapat pada dataset.

```python
print("Jumlah record :", df.shape[0])
print("Jumlah attribute :", df.shape[1])

print("\nNama attribute:")
for i, column in enumerate(df.columns, 1):
    print(f"{i}. {column}")
```

### **4. Melihat Tipe Data Setiap Attribute**

Setiap attribute memiliki tipe data yang berbeda. Pada tahap ini

```python
df.dtypes
```

### **6. Mengelompokkan Attribute Berdasarkan Tipe Data**

Selanjutnya attribute dipisahkan menjadi attribute bertipe angka dan
attribute bertipe object. Attribute angka akan digunakan untuk analisis
statistik dan korelasi.

```python
numeric_columns = df.select_dtypes(include='number').columns
object_columns = df.select_dtypes(include='object').columns

print("Jumlah attribute bertipe angka :", len(numeric_columns))
print("Jumlah attribute bertipe object :", len(object_columns))

print("\nAttribute bertipe angka:")
for column in numeric_columns:
    print("-", column)

print("\nAttribute bertipe object:")
for column in object_columns:
    print("-", column)
```

### **7. Sepuluh Record Pertama Attribute Angka**

Menampilkan  10 record pertama dengan hanya mengambil attribute yang bertipe angka.

```python
df[numeric_columns].head(10)
```

### **8. Statistik Deskriptif Attribute Angka**

Statistik deskriptif digunakan untuk mengetahui gambaran umum dari
data numerik. Nilai yang dicari meliputi rata-rata, standar deviasi,
nilai minimum, maksimum, Q1, Q2, dan Q3.

```python
statistik = pd.DataFrame({
    "Mean": df[numeric_columns].mean(),
    "Std Dev": df[numeric_columns].std(),
    "Min": df[numeric_columns].min(),
    "Q1": df[numeric_columns].quantile(0.25),
    "Q2": df[numeric_columns].quantile(0.50),
    "Q3": df[numeric_columns].quantile(0.75),
    "Max": df[numeric_columns].max()
})

statistik
```

### **9. Jumlah Label pada Attribute Object**

Attribute object berisi data berupa kategori atau teks. Pada bagian ini
dilihat jumlah masing-masing label dari seluruh attribute object.

```python
for column in object_columns:
    print(f"\n===== {column} =====")
    print(df[column].value_counts(dropna=False))
```

### **10. Pengecekan Missing Value**

Sebelum melakukan visualisasi dan analisis lebih lanjut, dilakukan
pengecekan terhadap data yang kosong pada setiap attribute.

```python
missing_value = pd.DataFrame({
    "Jumlah Missing": df.isnull().sum(),
    "Persentase Missing (%)": df.isnull().mean() * 100
})

missing_value
```

### **11. Pengecekan Data Duplikat**

Pengecekan ini dilakukan untuk mengetahui apakah terdapat record
yang sama atau tercatat lebih dari satu kali.

```python
print("Jumlah data duplikat :", df.duplicated().sum())
```

### **12. Visualisasi 1 — Distribusi Species**

Visualisasi pertama menggunakan bar chart. Grafik ini dipilih karena
sesuai untuk membandingkan jumlah data pada beberapa kategori.
Pada bagian ini yang dibandingkan adalah jumlah penguin berdasarkan
species.

```python
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x="Species")

plt.title("Jumlah Penguin Berdasarkan Species")
plt.xlabel("Species")
plt.ylabel("Jumlah Penguin")
plt.xticks(rotation=15)

plt.show()
```

Berdasarkan grafik, dapat dilihat bahwa jumlah data pada setiap species
tidak sama. Bar chart memudahkan untuk membandingkan jumlah masing-masing
species.

### **13. Visualisasi 2 — Distribusi Body Mass**

Visualisasi kedua menggunakan histogram. Histogram dipilih karena
digunakan untuk melihat penyebaran data numerik berdasarkan rentang nilai.
Attribute yang digunakan adalah Body Mass karena merupakan salah satu
ukuran numerik pada data penguin.

```python
plt.figure(figsize=(8, 5))

sns.histplot(data=df, x="Body Mass (g)", bins=20, kde=True)

plt.title("Distribusi Body Mass Penguin")
plt.xlabel("Body Mass (g)")
plt.ylabel("Jumlah Penguin")

plt.show()
```

Histogram menunjukkan bagaimana data Body Mass tersebar pada beberapa
rentang nilai. Dari grafik dapat dilihat bagian rentang massa tubuh
yang memiliki jumlah data lebih banyak.

### **14. Melihat Korelasi Attribute Numerik**

Setelah melihat statistik dan distribusi data, selanjutnya dilihat
hubungan antar attribute numerik menggunakan korelasi.

Korelasi digunakan untuk melihat apakah dua attribute memiliki hubungan
yang searah, berlawanan, atau hubungan yang relatif lemah.

```python
correlation = df[numeric_columns].corr()

correlation
```

### **15. Heatmap Korelasi**

Heatmap digunakan untuk menampilkan nilai korelasi antar attribute
numerik dalam bentuk warna. Nilai korelasi berada pada rentang -1
sampai 1.

```python
plt.figure(figsize=(11, 8))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Heatmap Korelasi Attribute Numerik")
plt.show()
```

### **16. Interpretasi Heatmap**

Pada heatmap, nilai yang mendekati 1 menunjukkan hubungan positif
yang kuat, sedangkan nilai yang mendekati -1 menunjukkan hubungan
negatif yang kuat. Nilai yang mendekati 0 menunjukkan hubungan linear
yang lebih lemah.

Dari heatmap dapat diamati attribute numerik mana yang memiliki
hubungan paling kuat dan mana yang memiliki hubungan lebih rendah.

### **17. Perbandingan Rata-rata Attribute Numerik Berdasarkan Species**

Selain melihat keseluruhan data, dilakukan perbandingan rata-rata
attribute numerik berdasarkan species. Hal ini dilakukan untuk

### **18. Perbandingan Body Mass Berdasarkan Species**

Untuk melihat perbedaan penyebaran Body Mass pada setiap species,
digunakan boxplot. Grafik ini dapat membantu melihat nilai tengah,
penyebaran data, dan nilai yang berbeda jauh dari data lainnya.

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="Species",
    y="Body Mass (g)"
)

plt.title("Perbandingan Body Mass Berdasarkan Species")
plt.xlabel("Species")
plt.ylabel("Body Mass (g)")
plt.xticks(rotation=15)

plt.show()
```
